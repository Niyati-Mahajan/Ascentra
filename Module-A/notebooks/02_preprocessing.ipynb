```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/data.csv", sep=";")

print("Dataset shape:", df.shape)
```

```python
df = df[df["Target"].isin(["Dropout", "Graduate"])].copy()

df["academic_risk"] = df["Target"].map({
    "Dropout": 1,
    "Graduate": 0
})

print(df["academic_risk"].value_counts())
```

```python
print(df.shape)
print(df["academic_risk"].value_counts())
```

```python
second_sem_columns = [
    col for col in df.columns
    if "2nd sem" in col
]

print("Columns to remove:")
for col in second_sem_columns:
    print(col)
```

```python
df.drop(columns=second_sem_columns, inplace=True)
```

```python
print("Remaining columns:", df.shape[1])
```

```python
X = df.drop(columns=["academic_risk"])
y = df["academic_risk"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_columns = X.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("\nNumber of numerical columns:", len(numerical_columns))
print("Number of categorical columns:", len(categorical_columns))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_columns
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_columns
        )
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)
```

```python
print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original test shape:", X_test.shape)
print("Processed test shape:", X_test_processed.shape)
```
