```python
def academic_risk_agent(student_profile):

    risk = student_profile["risk_probability"]
    level = student_profile["risk_level"]
    monitoring = student_profile["monitoring_data"]

    actions = []
    priorities = []

    if level == "HIGH":
        priorities.append("Immediate academic intervention")
    elif level == "MEDIUM":
        priorities.append("Early academic intervention")
    else:
        priorities.append("Routine monitoring")

    if monitoring["attendance_percentage"] < 75:
        actions.append(
            "Monitor attendance and identify barriers to regular class participation."
        )

    if monitoring["best_2_ca_average"] < 15 or monitoring["mid_term_score"] < 15:
        actions.append(
            "Provide targeted revision support for continuous assessments and the mid-term exam."
        )

    if monitoring["ca1_score"] < 12 or monitoring["ca2_score"] < 12 or monitoring["ca3_score"] < 12:
        actions.append(
            "Review CA-wise performance and set a short improvement plan before the next assessment."
        )

    previous_tgpa = monitoring.get("previous_semester_tgpa")
    if previous_tgpa == previous_tgpa and previous_tgpa < 5.0:
        actions.append(
            "Discuss previous-semester academic gaps and plan subject-level support."
        )

    if monitoring["academic_trend"] == "Declining":
        actions.append(
            "Schedule an academic advisor follow-up to review the declining trend."
        )

    return {
        "risk_probability": risk,
        "risk_level": level,
        "priority": priorities,
        "recommended_actions": actions
    }
```

```python
import pandas as pd
import numpy as np
import joblib
import shap
import json

df = pd.read_csv("../data/ascentra_student_data_refined.csv")
target_column = "academic_risk" if "academic_risk" in df.columns else df.columns[-1]

xgb_ascentra = joblib.load(
    "../models/ascentra_xgboost.pkl"
)

preprocessor = joblib.load(
    "../models/ascentra_preprocessor.pkl"
)

explainer = shap.TreeExplainer(xgb_ascentra)

print("Everything loaded successfully!")
```

```python
def generate_student_risk_profile(student_row):

    target_column = "academic_risk" if "academic_risk" in student_row.index else None
    drop_labels = ["student_id"] + ([target_column] if target_column else [])

    X_student = student_row.drop(
        labels=drop_labels,
        errors="ignore"
    ).to_frame().T

    X_processed = preprocessor.transform(X_student)

    risk_probability = xgb_ascentra.predict_proba(
        X_processed
    )[0][1]

    prediction = xgb_ascentra.predict(
        X_processed
    )[0]

    shap_values = explainer.shap_values(
        X_processed
    )

    if isinstance(shap_values, list):
        shap_values = shap_values[0]

    shap_values = np.asarray(shap_values).reshape(-1)

    feature_names = preprocessor.get_feature_names_out()

    shap_df = pd.DataFrame({
        "Feature": feature_names,
        "SHAP Value": shap_values
    })

    shap_df["Absolute SHAP"] = shap_df["SHAP Value"].abs()

    shap_df = shap_df.sort_values(
        "Absolute SHAP",
        ascending=False
    )

    risk_increasing = shap_df[shap_df["SHAP Value"] > 0].head(5)
    risk_reducing = shap_df[shap_df["SHAP Value"] < 0].sort_values("SHAP Value").head(5)

    if risk_probability >= 0.70:
        risk_level = "HIGH"
    elif risk_probability >= 0.40:
        risk_level = "MEDIUM"
    else:
        risk_level = "LOW"

    monitoring_features = [
        "attendance_percentage",
        "ca1_score",
        "ca2_score",
        "ca3_score",
        "best_2_ca_average",
        "mid_term_score",
        "previous_semester_tgpa",
        "academic_trend"
    ]

    monitoring_data = student_row[monitoring_features]

    return {
        "student_id": student_row["student_id"],
        "model_prediction": int(prediction),
        "risk_probability": round(float(risk_probability * 100), 2),
        "risk_level": risk_level,
        "monitoring_data": monitoring_data.to_dict(),
        "risk_increasing_factors": risk_increasing[["Feature", "SHAP Value"]].to_dict("records"),
        "risk_reducing_factors": risk_reducing[["Feature", "SHAP Value"]].to_dict("records")
    }
```

```python
student = df.iloc[0]

student_profile = generate_student_risk_profile(
    student
)

print(
    json.dumps(
        student_profile,
        indent=4
    )
)
```

```python
agent_result = academic_risk_agent(
    student_profile
)

print(
    json.dumps(
        agent_result,
        indent=4
    )
)
```

```python
def generate_intervention_plan(student_profile):

    monitoring = student_profile["monitoring_data"]
    risk = student_profile["risk_probability"]
    level = student_profile["risk_level"]

    interventions = []

    if monitoring["attendance_percentage"] < 75:
        interventions.append({
            "area": "Attendance",
            "priority": "HIGH",
            "action": "Improve class attendance and identify barriers affecting participation."
        })

    if monitoring["best_2_ca_average"] < 15 or monitoring["mid_term_score"] < 15:
        interventions.append({
            "area": "Assessment Performance",
            "priority": "HIGH",
            "action": "Provide targeted revision support for continuous assessments and the mid-term exam."
        })

    if monitoring["ca1_score"] < 12 or monitoring["ca2_score"] < 12 or monitoring["ca3_score"] < 12:
        interventions.append({
            "area": "Continuous Assessment",
            "priority": "MEDIUM",
            "action": "Review CA-wise scores and set an assessment recovery plan."
        })

    previous_tgpa = monitoring.get("previous_semester_tgpa")
    if previous_tgpa == previous_tgpa and previous_tgpa < 5.0:
        interventions.append({
            "area": "Previous Semester Performance",
            "priority": "MEDIUM",
            "action": "Discuss previous-semester gaps and arrange subject-level academic support."
        })

    if monitoring["academic_trend"] == "Declining":
        interventions.append({
            "area": "Academic Trend",
            "priority": "HIGH",
            "action": "Schedule an academic advisor follow-up to investigate the declining trend."
        })

    if level == "HIGH":
        overall_action = "Immediate academic intervention and advisor follow-up recommended."
    elif level == "MEDIUM":
        overall_action = "Early intervention and continued monitoring recommended."
    else:
        overall_action = "Continue routine monitoring and provide preventive support."

    return {
        "student_id": student_profile["student_id"],
        "risk_probability": risk,
        "risk_level": level,
        "overall_action": overall_action,
        "interventions": interventions
    }
```

```python
def agent_decision(student_profile, intervention_plan):

    risk_probability = student_profile["risk_probability"]
    risk_level = student_profile["risk_level"]
    monitoring = student_profile["monitoring_data"]

    decisions = []

    if monitoring["attendance_percentage"] < 75:
        decisions.append({
            "area": "Attendance",
            "priority": "HIGH",
            "decision": "Immediate attendance intervention",
            "action": "Advisor meeting and weekly attendance monitoring"
        })

    if (
        monitoring["best_2_ca_average"] < 15
        or monitoring["mid_term_score"] < 15
        or monitoring["ca1_score"] < 12
    ):
        decisions.append({
            "area": "Academic Performance",
            "priority": "HIGH",
            "decision": "Academic support required",
            "action": "Provide tutoring, revision support, and monitor CA/mid-term performance"
        })

    previous_tgpa = monitoring.get("previous_semester_tgpa")
    if previous_tgpa == previous_tgpa and previous_tgpa < 5.0:
        decisions.append({
            "area": "Previous Semester Performance",
            "priority": "MEDIUM",
            "decision": "Review historical academic gaps",
            "action": "Plan subject-specific support based on previous-semester performance"
        })

    if monitoring["academic_trend"] == "Declining":
        decisions.append({
            "area": "Academic Trend",
            "priority": "HIGH",
            "decision": "Escalate academic monitoring",
            "action": "Advisor follow-up and bi-weekly progress review"
        })

    if risk_probability >= 75:
        overall_decision = "ESCALATE"
    elif risk_probability >= 50:
        overall_decision = "MONITOR"
    else:
        overall_decision = "ROUTINE"

    return {
        "student_id": student_profile["student_id"],
        "risk_probability": risk_probability,
        "risk_level": risk_level,
        "overall_decision": overall_decision,
        "decisions": decisions
    }
```

```python
def generate_agent_response(student_profile, agent_decision_result):

    prompt = f"""
You are ASCENTRA's Academic Intervention Agent.

The machine-learning system has already calculated the
student's risk. You must NOT change that prediction.

Your task is to convert the agent's decisions into a
clear and actionable intervention plan.

STUDENT PROFILE:
{json.dumps(student_profile, indent=2)}

AGENT DECISIONS:
{json.dumps(agent_decision_result, indent=2)}

Provide:

1. Immediate Action
2. Priority Interventions
3. Student Support Strategy
4. Advisor Follow-up
5. Monitoring Metrics
6. Escalation Condition

Rules:
- Do not invent student information.
- Do not modify the ML risk probability.
- Prioritize high-risk issues.
- Give practical actions rather than generic advice.
- Keep the response concise and professional.
"""

    response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt
)

    return response.text
```

```python
import json

def predict_student_risk(student_profile):

    # -----------------------------
    # 1. Generate intervention plan
    # -----------------------------
    intervention_plan = generate_intervention_plan(
        student_profile
    )

    # -----------------------------
    # 2. Agent decision
    # -----------------------------
    agent_decision_result = agent_decision(
        student_profile,
        intervention_plan
    )

    # -----------------------------
    # 3. Create structured report
    # -----------------------------
    agent_report = create_agent_report(
        student_profile,
        intervention_plan
    )

    # -----------------------------
    # 4. Generate Gemini response
    # -----------------------------
    ai_response = generate_agent_response(
        student_profile,
        agent_decision_result
    )

    # -----------------------------
    # 5. Final Ascentra output
    # -----------------------------
    final_result = {
        "student_id": student_profile["student_id"],
        "risk_probability": student_profile["risk_probability"],
        "risk_level": student_profile["risk_level"],
        "agent_decision": agent_decision_result,
        "intervention_plan": intervention_plan,
        "ai_advisor_response": ai_response
    }

    return final_result
```
